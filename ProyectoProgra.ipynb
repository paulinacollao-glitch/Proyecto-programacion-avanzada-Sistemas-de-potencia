```python
class ElementoRed:
  """Clase base para representar un elemento de la red eléctrica."""
  def __init__(self, id_equipo):
    """Inicializa un elemento de la red con un identificador único.
    
    args:
        id_equipo (str): Identificador único del equipo.
    """
    self.id_equipo=id_equipo

class Generador(ElementoRed):
  """representa un generador eléctrico en la red."""
  def __init__(self, id_equipo,potencia,potencia_max, costo_op):
    """Inicializa un generador con sus propiedades específicas.
    args:
        id_equipo (str): Identificador único del generador.
        potencia (float): Potencia actual del generador en MW.
        potencia_max (float): Potencia máxima que puede generar el generador en MW.
        costo_op (float): Costo operativo del generador por unidad de energía producida.
    """
    super().__init__(id_equipo)
    self.__potencia=potencia
    self.__potencia_max=potencia_max
    self.__costo_op=costo_op

  def get_potencia(self):
    """Obtiene la potencia actual del generador.
    returns:
        float: Potencia actual del generador en MW.
    """
    return self.__potencia

  def get_potencia_max(self):
        """Obtiene la potencia máxima del generador.
    returns:
        float: Potencia máxima del generador en MW.
    """
        return self.__potencia_max

  def set_potencia(self, potencia):
    """Establece la potencia actual del generador.
    args:
        potencia (float): Nueva potencia del generador en MW.
    """
    if potencia <= self.get_potencia_max():
      self.__potencia=potencia
    else:
      print(f"Error: La potencia {potencia} MW excede la potencia máxima {self.get_potencia_max()} MW del generador {self.id_equipo}.")


  def get_costo_op(self):
    return self.__costo_op

class LineaTransmision(ElementoRed):
  def __init__(self, id_equipo,corriente,corriente_max,impedancia):
    super().__init__(id_equipo)
    self.__corriente=corriente
    self.__corriente_max=corriente_max
    self.impedancia=impedancia

  def get_corriente(self):
    return self.__corriente
  def get_corriente_max(self):
    return self.__corriente_max

class Carga(ElementoRed):
  def __init__(self, id_equipo,demanda):
    super().__init__(id_equipo)
    self.__demanda=demanda

  def get_demanda(self):
    return self.__demanda
```

```python
#clase gestora
class SistemaPotencia:
  def __init__(self):
    self.elementos={}
  def agregar_elemento(self, elemento):
      self.elementos[elemento.id_equipo] = elemento

  def buscar_elemento_por_id(self, id_equipo):
      return self.elementos.get(id_equipo, None)
  
  def _calcular_total(self, tipo_clase, metodo_getter):# evita DRY - Don't Repeat Yourself
        """Método auxiliar genérico para sumar valores según el tipo de elemento."""
        total = 0
        for elemento in self.elementos.values():
            if isinstance(elemento, tipo_clase):
            # getattr para llamar dinámicamente al método getter (ej: 'get_demanda')
                total += getattr(elemento, metodo_getter)()
        return total
  def calcular_costo_total(self):
      return self._calcular_total(Generador, "get_costo_op")

  def calcular_demanda_total(self):
      return self._calcular_total(Carga, "get_demanda")

  def calcular_corriente_total(self):
      return self._calcular_total(LineaTransmision, "get_corriente")
```
